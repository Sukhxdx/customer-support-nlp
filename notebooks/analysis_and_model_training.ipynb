# Customer Support NLP — Analysis & Model Training

Exploratory analysis and model training for the Customer Support Ticket Classification project.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from src.data_preprocessing import load_data, preprocess_dataframe, get_category_distribution
from src.model_training import compare_models, split_train_test, save_best_model
from src.evaluation import save_confusion_matrix, save_model_comparison_chart, print_classification_report
```

```python
df = load_data()
df = preprocess_dataframe(df)
print(f"Shape: {df.shape}")
df.head()
```

```python
dist = get_category_distribution(df)
pd.Series(dist).plot(kind="bar", title="Category Distribution", color="steelblue")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
X_train, X_test, y_train, y_test = split_train_test(df)
comparison_df, best_name, best_pipeline, _ = compare_models(X_train, X_test, y_train, y_test)
print(comparison_df)
print(f"\nBest model: {best_name}")
```

```python
y_pred = best_pipeline.predict(X_test)
save_best_model(best_pipeline, comparison_df, best_name, y_test, y_pred)
print_classification_report(y_test, y_pred)
save_confusion_matrix(y_test, y_pred)
save_model_comparison_chart(comparison_df)
```
